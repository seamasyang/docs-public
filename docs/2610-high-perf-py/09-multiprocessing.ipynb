同步： 进程间同步控制流程锁和信号量

| 对比项 | 线程之间的锁 | 进程之间的锁 |
|---|---|---|
| 主要用途 | 保护多线程共享资源 | 保护多进程共享资源 |
| 典型 API | `threading.Lock()` | `multiprocessing.Lock()` |
| 作用范围 | 同一进程内的线程 | 共享同一个 `Lock` 对象的进程 |
| 内存空间 | 线程共享进程内存 | 进程通常隔离内存 |
| 同步开销 | 通常较低 | 通常较高 |
| 适用场景 | 多线程修改共享变量 | 多进程访问共享资源、文件或设备 |
| 能否互相替代 | 不能直接替代 | 不能直接替代 |

threading.Lock → 同一进程内的线程。


multiprocessing.Lock → 多个进程。


asyncio.Lock → 同一事件循环中的协程。

| 机制 | 控制什么 | 示例 |
|---|---|---|
| Semaphore | 同一时刻有多少任务在执行 | 最多 10 个并发请求 |
| Rate Limiter | 一段时间内允许多少次请求 | 每分钟最多 100 次请求 |
| Queue | 任务如何排队等待执行 | GPU 推理任务依次排队 |


| 使用场景 | 计数器的共享范围 |
|---|---|
| `threading.Semaphore(3)` | 使用同一个信号量对象的线程 |
| `multiprocessing.Semaphore(3)` | 共享同一个信号量对象的进程 |
| `asyncio.Semaphore(3)` | 使用同一个信号量对象的协程 |
