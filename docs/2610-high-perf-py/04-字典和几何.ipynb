## key is hashable, support： int, str, tuple; 
## key is unique
## value can be any type

```python
# dict； key-value pair
d = {"name":"alice", "age":30}

d.keys() # ["name", "age"]
d.values() # ["alice", 30]
d.items() # [("name","alice"), ("age",30)]

d.update({"dept":"it"})
print(d)

d.update({"dept":"hr"})
print(d)

last = d.popitem()
print(last)
print(d)


```

```python
counts = {}
for x in [1, 1, 2, 3, 3, 3]:
    counts[x] = counts.get(x, 0) + 1

print(counts)
```

元素唯一，自动去重；

元素必须可哈希；

**无序**（不要依赖顺序）；

查找、插入、删除平均 O(1)；

支持集合运算；

**不能索引**：s[0] 会报错。

```python
# set, 不重复
s= {1, 2, 3, 5}

s.add(4)
print(s)

s.remove(4)
print(s)

s.discard(5)
print(s)

item = s.pop()
print(item)
print(s)

```

```python
a = {1, 2, 3}
b = {3, 4, 5}



print("union:", a|b) 
print("intersection:",a&b)
print("a difference b:",a-b)
print("b difference a:",b-a)
print("symmetric_difference:",a^b)
```
