```python
# % python -m pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu128

import torch

print("PyTorch version:", torch.__version__)
print("PyTorch CUDA build:", torch.version.cuda)
print("CUDA available:", torch.cuda.is_available())

try:
    print("Compiled CUDA:", torch.backends.cuda.is_built())
    print("GPU count:", torch.cuda.device_count())
    print("CUDA error:", torch.cuda.init() or "None")
except Exception as e:
    print("CUDA initialization failed:", repr(e))
```

```python

import time
import torch

if not torch.cuda.is_available():
    raise RuntimeError("CUDA is unavailable. Check your PyTorch installation.")

device = torch.device("cuda")
print("GPU:", torch.cuda.get_device_name(0))
print("VRAM:", round(
    torch.cuda.get_device_properties(0).total_memory / 1024**3, 2
), "GiB")

# Adjust this to control GPU memory usage.
size = 4096

a = torch.randn(size, size, device=device)
b = torch.randn(size, size, device=device)

print("Running GPU matrix multiplication...")
start = time.perf_counter()

try:
    while True:
        # Large matrix multiplications keep the GPU busy.
        c = torch.mm(a, b)
        a = torch.relu(c)
        b = torch.mm(a, c)
        torch.cuda.synchronize()

        elapsed = time.perf_counter() - start
        print(f"\rElapsed: {elapsed:.1f}s", end="", flush=True)

except KeyboardInterrupt:
    print("\nStopped.")

finally:
    del a, b
    if "c" in locals():
        del c
    torch.cuda.empty_cache()
```
