```python
# 斐波那契（无限序列）

def fib():
    a, b = 1, 2
    while True:
        yield a
        a, b = b, a + b

f = fib()
print([next(f) for _ in range(10)])
```

```python
# 分批读取

def chunks(iterable, size):
    batch = []
    for x in iterable:
        batch.append(x)
        if len(batch) == size:
            yield batch
            batch = []
    if batch:
        yield batch

list(chunks(range(10), 3))
```
